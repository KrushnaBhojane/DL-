```python
import tensorflow as tf 

a = tf.constant(5)
b = tf.constant(3)
print("Tensorflow Addition", (a+b).numpy())
```

```python
from tensorflow import keras 

model = keras.Sequential([
    keras.Input(shape=(1,)),
    keras.layers.Dense(1)
])
print("Keras Model is created")
model.summary()
```

```python
import theano 
import theano.tensor as T 
x = T.dscalar("x")
y = T.dscalar("y")
add = theano.function([x,y], x+y)
print("Theano Addition:", add(5,10))
```

```python
import torch

x = torch.tensor(5)
y = torch.tensor(10)
print("Pytorch Addition:", (x+y).item())
```
